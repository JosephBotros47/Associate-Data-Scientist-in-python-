```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Histogram :A visual representation of the distribution of quantitative data. 
#            They group values into bins (intervals)
#            and counts how many fall in each range.
```

```python
scores = np.random.normal(loc= 80 ,scale = 10, size = 100)# loc = median , scale = standard deviation , size of maximum number = 100
```

```python
plt.hist(scores)
```

```python
scores = np.random.normal(loc= 100 ,scale = 20, size = 200)
plt.hist(scores)
```

```python
# change number of bins
scores = np.random.normal(loc= 100 ,scale = 20, size = 200)
plt.hist(scores , bins = 10)
```

```python
plt.hist(scores , bins = 20)
```

```python
plt.hist(scores , bins = 30)
```

```python
# change the color and edge color 
plt.hist(scores , bins = 30 , color = "green", edgecolor = "black")
```
