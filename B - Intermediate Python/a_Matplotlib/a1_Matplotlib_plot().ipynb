```python
import matplotlib.pyplot as plt
```

```python
x = [2023, 2024, 2025, 2026]
y = [15 , 25, 30 , 20]

plt.plot(x,y)
plt.show()
```

```python
import numpy as np
x_array = np.array(x)
y_array = np.array(y)
plt.plot(x_array,y_array)
```

```python
# Plot Customization 
import numpy as np
x_array = np.array(x)
y_array = np.array(y)
plt.plot(x_array,y_array)
```

```python
# to mark the points 
plt.plot(x_array, y_array, marker = ".")
```

```python
plt.plot(x_array, y_array, marker = "o")
```

```python
# To change the marker size 
plt.plot(x_array, y_array, marker = "o" ,markersize = 30)
```

```python
# We can make markersize = ms to simple it 
plt.plot(x_array, y_array, marker = "o" ,ms = 20)
```

```python
# To change the marker color 
plt.plot(x_array, y_array, marker = "o" , ms = 20 , markerfacecolor = '#1cd3fc')
```

```python
# You can sample it to mfc
plt.plot(x_array, y_array, marker = "o", ms = 20 , mfc = '#1cd3fc')
```

```python
# To change the line color
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc')
```

```python
# TO Change the line style  ....
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dotted")
```

```python
# TO Change the line style  -.-.-.-.
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot")
```

```python
# TO Change the line style ______
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "solid")
```

```python
# TO Change the line width 
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 2)
```

```python
# TO Change the line width 
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 4)
```

```python
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 4)
y2_array = np.array([17,23,38,5])
plt.plot(x_array, y2_array)
```

```python
plt.plot(x_array,y_array, marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 4)
y2_array = np.array([17,23,38,5])
plt.plot(x_array, y2_array ,marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 4)
```

```python
# You can take your attributes af your style into dictionary 
my_style = dict(marker = '.' , ms = 20, mfc = '#1cd3fc' , markeredgecolor = '#1cd3fc',
        linestyle = "dashdot", linewidth= 4)
# To access this dictionary use ** before name of dictionary
plt.plot(x_array, y_array ,**my_style)
plt.plot(x_array, y2_array ,**my_style)
```
