#**Subsetting Rows**

```python
df["Gross Amount"] < 142
```

```python
low = df[df["Gross Amount"] < 142]
low
```

```python
category = df[df["Product Category"] == "Clothing"]
category
```

```python
# Subsetting based on multiple conditions
is_closing = df["Product Category"] == "Clothing"
is_female = df['Gender'] == "Male"
clothing_Male = df[is_closing & is_female]
clothing_Male
```

```python
# Subsetting using .isin()
categories = df["Product Category"].isin(["Electronics","Clothing"])
categories
```
