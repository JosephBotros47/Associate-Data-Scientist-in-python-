As you can see here, the flow is straightforward: data lives in a source (a database, cloud storage, etc.), gets exported to a data file like a CSV, and then Python loads it into a DataFrame to produce insights.

### **Data Source  ->  Data files  ->  R or Python  -> Insights**

```python

"""The syntax E_commerse['Gross Amount'] on line 1 is how you pull any column out of
 a DataFrame — just put the column name in quotes inside square brackets.
 The result gets stored in the variable amounts """
amounts = E_commerce['Gross Amount']
amounts
```

```python
from numpy._core.fromnumeric import mean
"""1 - Data Aggregation"""
# This pattern — target.method(parameters) —
# is consistent across all pandas operations, so it's worth having firmly in mind.

# 1 - What's the Total order amount? Sum()
revenue = amounts.sum()
print(revenue)

# 2 - What's the smallest order amount? → min()
minimum_amount = amounts.min()
print(minimum_amount)

# 3 - What's the largest order amount? → max()
maximum_amount = amounts.max()
print(maximum_amount)

# 4 - What's the average order amount? → mean()
mean_amount = amounts.mean()
print(mean_amount)
```

```python
"""The Difference between median and mean"""
# Mean : sum of all values divided by the count. Every value contributes, including extreme ones.

# 5 - Median :  the middle value when data is sorted. It ignores how extreme the outliers are.

"""
When to use each:

Use median :
 when outliers could distort the picture (e.g., income, house prices
 , order amounts with a few huge corporate orders)


Use mean when there are no significant outliers, or when you specifically want
all values — including extremes — to influence the result
"""

# if the mean was close to median which tells us there aren't major outliers skewing the data in this dataset
median_amount = amounts.median()
print(median_amount)
```

```python
# 6 - Variance .var()

# 7 - Standard deviation .std()
```

```python
""" Comulative statistics """
# that the operation be on the current value and the previos values
# .cum
# .cumsum()  -> that the result of any value is its value + all previos values
print(amounts.cumsum())

# .cummin()  ->
print(amounts.cummin()) # the minimum of the series from begin to the index you give

# .cummax()  ->
print(amounts.cummax()) # the maximim of the series from begin to the index you give

# .cumprod()  -> that the result of any value is its value * all previos values
print(amounts.cumprod())
```

```python
"""What Does .agg() Do in Pandas?"""
# The .agg() (short for aggregate) method in Pandas is used to compute summary statistics
# on your DataFrame or Series. It provides flexibility by allowing you to apply custom functions,
# calculate statistics across multiple columns, or run multiple summary operations all at once.

## using many of basic operation in one call
print(amounts.agg(['min', 'max', 'mean']))

# using a custom operation(function) in one call
# A custom IQR function
def iqr(column):
    return column.quantile(0.75) - column.quantile(0.25)

# Print IQR of the amounts
print(amounts.agg(iqr))

## Using it for multiple columns
print(E_commerce[["Gross Amount","Discount Amount (INR)","Net Amount"]].agg(iqr))
print(E_commerce[["Gross Amount","Discount Amount (INR)","Net Amount"]].agg(["max", "mean"]))
print(E_commerce[["Gross Amount","Discount Amount (INR)","Net Amount"]].agg(["max", "mean",iqr]))
```

```python
"""
Whenever you design a data aggregation, answer two questions:

Which column do we wish to summarize? — For total revenue,
it was amount. For total distance, it was distance. The column
should contain the values relevant to your question.
Which summary operation do we wish to apply? — As you can see here,
the same column can feed into sum, mean, min, or max depending on what insight you need.
These two questions — column and operation — are all you need to design
any aggregation. The rest is just syntax.
"""
```

```python
# count()
# count() counts all non-null values in the column, giving us the total number of rows
Transform_count = E_commerce['TID'].count()
print(Transform_count)

# nunique()
# nunique() scans the user_id column, identifies all distinct values, and counts
# only those — duplicates are ignored entirely.
customer_count = E_commerce['CID'].nunique()
print(customer_count)

# counting
# value_count()
E_commerce["Purchase Method"].value_counts()

# value_count(sort = True)
E_commerce["Purchase Method"].value_counts(sort = True)
# show me the values with sorting
```
