# **Explicit Indexes**

```python
import pandas as pd
df = pd.read_csv("/content/project1_df.csv")
```

```python
df.index
```

```python
df.columns
```

```python
# Setting a column as the index
df_index = df.set_index("CID")
```

```python
df_index.head(2)
```

```python
# reset the index
df_index.reset_index()
```

```python
df_index = df.set_index("CID")
```

```python
df_index.head(2)
```

```python
# Dropping an index
df_index.reset_index(drop = True)
```

```python
df_index = df.set_index("TID")
```

```python
# Indexes make subsetting simpler
df_index.loc[[5876328741,5876328741 ]]
```

```python
# Index values don't need to be unique
df_index.reset_index()
df_index = df_index.set_index("Product Category")
```

```python
# Subsetting on duplicated index values
df_index.loc["Electronics"]
```

```python
df_index.reset_index()
```

```python
# Multi-level indexes a.k.a. hierarchical indexes
df = df.set_index(["Gender","Location"])
```

```python
df.loc[["Male","Mumbai"]]
```
